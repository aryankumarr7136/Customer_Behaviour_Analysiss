```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df =df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
#creating new column 
labels = ['Young Adult' ,'Adult','middle Aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4 , labels=labels)
```

```python
df[['age' ,'age_group']].head(10)
```

```python
# create column purchase frequency days
frequency_mapping = {
    'Fortnightly' :14,
    'Weekly' : 7,
    'Monthly':30,
    'Quaterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
user = "postgres"
password = "1234"
host = "localhost"  # or your database host IP
port = "5432"  # default PostgreSQL port
database_name = "customer_behaviour"

# Create the database connection string / engine
engine = create_engine(
    f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{database_name}"
)
# Write the DataFrame to a table in PostgreSQL
table_name = "customer"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",  # Options: 'fail', 'replace', 'append'
    index=False,  # Excludes the Pandas index column
)

print(f"Successfully loaded {len(df)} rows into table '{table_name}'.")
```
